## 3. Success metric

*One metric you can defend. What number means 'good'?*

The main success metric will be Precision@K, especially Precision@50.

Precision@50 measures how many of the top 50 pages selected by the ranking are actually positive according to the provisional target.

This metric matches the real decision because a reviewer has limited time and will normally review a limited number of pages first.

I will also consider Average Precision and ROC AUC as supporting metrics, but Precision@K is the most directly connected to the review-queue use case.

A successful model should rank more relevant pages near the top than a simple baseline rule.

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

A fixed rule can identify simple patterns, such as pages with low CTR, low engagement, old content, or declining trends. However, a fixed rule usually requires manually choosing weights and thresholds for each signal.

ML can evaluate multiple observable signals together and learn which combinations are more useful for prioritization.

For example, a page may have a downward trend but also have high impressions, strong search visibility, and a position that makes the opportunity more important than another declining page with very little demand.

The goal is not to replace simple rules. The rule-based approach provides a transparent baseline, while ML can be tested against it to determine whether it produces a better ranked review queue.

If ML does not improve the ranking performance over the baseline, the simpler rule may be preferable.
